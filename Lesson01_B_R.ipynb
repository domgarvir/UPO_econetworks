## Graphs Methods and NetworkX functions  ➜  igraph methods/functions in R

In Python you used **NetworkX** graph *methods* (e.g., `G.nodes()`, `G.edges()`, `G.has_edge(...)`, `G.degree(...)`) and some **NetworkX functions** (e.g., `nx.is_connected`).

In R, the closest equivalents are mostly in **igraph**.

We'll use:
- `V(g)` for vertices, `E(g)` for edges
- `vcount(g)`, `ecount(g)` for counts
- `are.connected(g, v, w)` and `%in% V(g)$name` for existence checks
- `degree(g)` for node degree
- `is_connected(g)` for connectivity
- `is_bipartite(g)` for bipartite check


### Setup
Load the packages used in this lesson.

```r
# If needed (run once):
# install.packages(c("igraph","tidyverse","tidygraph","ggraph"))

library(igraph)
library(tidyverse)
library(tidygraph)
library(ggraph)
```

### Graph Methods

In this part, you will practice the R/igraph equivalents of common NetworkX graph methods.

To keep this notebook self-contained, we'll build a small example network. Later, we can replace it with your course dataset (e.g., the *Lord of the Rings* network) once we port `load_LotR_network()`.


```r
# Example network (replace later with your course loader)
edges <- tribble(
  ~from, ~to,
  "frodo", "sam",
  "frodo", "gandalf",
  "sam",   "merry",
  "sam",   "pippin",
  "gandalf","aragorn",
  "aragorn","legolas",
  "aragorn","gimli"
)

g <- graph_from_data_frame(edges, directed = FALSE)
g
```

#### Obtaining nodes and edges in the network

### Exercise 1
1. List all node names.
2. List all edges.
3. Iterate over nodes and print them.

**NetworkX hint:** `G.nodes()`, `G.edges()`, `for n in G.nodes(): ...`


```r
# 1) Nodes
V(g)$name

# 2) Edges (as a data frame)
as_data_frame(g, what = "edges") |> as_tibble()

# 3) Iterate
for (n in V(g)$name) {
  print(n)
}
```

#### Checking for existence of nodes and links

### Exercise 2
1. Check whether a node exists (e.g., "tomb" or any name you choose).
2. Check whether an edge exists between two nodes.

**NetworkX hint:** `G.has_node(u)`, `G.has_edge(u, v)`


```r
# 1) Node existence
"tomb" %in% V(g)$name

# 2) Edge existence (undirected example)
are.connected(g, "frodo", "sam")
are.connected(g, "frodo", "legolas")
```

#### Node degree

### Exercise 3
1. Compute the degree of every node.
2. Extract the degree of a single node (e.g., "sam").
3. Identify the node(s) with maximum degree.

**NetworkX hint:** `G.degree()`


```r
deg <- degree(g)
deg

deg["sam"]

max_deg <- max(deg)
names(deg[deg == max_deg])
```

#### Basic size properties

### Exercise 4
Compute the number of nodes and edges.

**NetworkX hint:** `G.number_of_nodes()`, `G.number_of_edges()`


```r
N <- vcount(g)
L <- ecount(g)

N
L
```

### NetworkX functions  ➜  igraph functions

In Python you used top-level functions like:
- `nx.is_connected(G)`
- `nx.is_bipartite(G)`
- `nx.draw(G)`

In R/igraph:
- connectivity: `is_connected(g)` (for undirected)
- bipartite: `is_bipartite(g)`
- plotting: `plot(g)` or `ggraph(...)`


#### Connectivity

### Exercise 5
1. Check if the graph is connected.
2. Make it disconnected by removing an edge or adding an isolated node.
3. Check connectivity again.


```r
is_connected(g)

# Make disconnected (add isolated node)
g_disc <- add_vertices(g, 1, name = "isolated")

is_connected(g_disc)
```

#### Bipartite check

### Exercise 6
1. Build a bipartite graph from an incidence matrix.
2. Check that it is bipartite.


```r
M <- matrix(c(
  1,0,1,
  0,1,1
), nrow = 2, byrow = TRUE)

rownames(M) <- c("Plant1","Plant2")
colnames(M) <- c("Poll1","Poll2","Poll3")

gB <- graph_from_incidence_matrix(M, weighted = TRUE)

is_bipartite(gB)
table(V(gB)$type)  # sets
```

#### Drawing graphs

### Exercise 7
Plot the graph using **ggraph**.


```r
set.seed(1)
as_tbl_graph(g) |>
  ggraph(layout = "fr") +
  geom_edge_link(alpha = 0.5) +
  geom_node_point(size = 3) +
  geom_node_text(aes(label = name), repel = TRUE, size = 3) +
  theme_void()
```
